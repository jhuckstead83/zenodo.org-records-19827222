## What this project studies

This release accompanies the preprint:

**Phase-Locked Extreme Compression in Adjacent Zeta-Zero Spacing: A Seam-Stitched Empirical Study Through the Full Available LMFDB Horizon**

The project studies **rare extreme compressions** in adjacent spacing between consecutive nontrivial zeros of the Riemann zeta function on the critical line.

The evidence chain in the current release is:

1. **Broad-board resolution ablation**  
   130k, 90k, 70k, and 52k boards retain the same dominant organizing coordinate.

2. **Terminal-band ladder**  
   The final 28.5B to 31.5B segment is compared directly at 52k, 8k, and 1k resolution.

3. **Matched-packet null**  
   The terminal 8k packet field exceeds a morphology-matched null ensemble.

4. **Jackknife robustness**  
   The terminal 8k lock remains stable after repeated random packet removal.


## What is claimed

The current release supports an **observed-horizon empirical claim**:

- rare extreme compressions organize reproducibly in the first logarithmic harmonic
- this organization remains stable under substantial row-size ablation
- the terminal band exhibits unusually tight local `h1` locking
- the terminal concentration is not explained by packet morphology alone or by a small outlier subset

### What is not claimed

This release does **not** claim:

- a closed asymptotic law beyond the observed horizon
- a proof of the Riemann Hypothesis
- that the fitted continuation family is a theorem at arbitrarily greater heights
